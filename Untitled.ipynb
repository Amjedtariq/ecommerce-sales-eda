```python
"""
=================================================================
تحليل بيانات مبيعات التجارة الإلكترونية  - EDA
=================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

```

```python
# ============================================================================
# الإعدادات
# ============================================================================
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
sns.set_style("whitegrid")

COLORS = ['#2E86AB', '#A23B72', '#F18F01', '#C73E1D', '#3B1F2B',
          '#5C9EAD', '#E84855', '#FFD23F', '#6A994E', '#BC4749']
sns.set_palette(COLORS)

os.makedirs('charts', exist_ok=True)

print("=" * 70)
print(" تحليل بيانات التجارة الإلكترونية ")
print("=" * 70)
```

```python

# ============================================================================
# 1) تحميل البيانات
# ============================================================================
print("\n📥 [1] تحميل البيانات...")
df = pd.read_csv('realistic_e_commerce_sales_data.csv')

# إزالة عمود الفهرس القديم
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

print(f"   ✅ شكل البيانات: {df.shape}")
print(f"   ✅ الأعمدة: {df.columns.tolist()}")
```

```python

# ============================================================================
# 2) الفهم الأولي للبيانات
# ============================================================================
print("\n🔍 [2] الفهم الأولي للبيانات...")
print("\n--- أول 5 صفوف ---")
print(df.head())

print("\n--- أنواع البيانات ---")
print(df.dtypes)

print("\n--- إحصائيات وصفية ---")
print(df.describe())

print("\n--- القيم المفقودة ---")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing': missing, 'Percentage': missing_pct})
print(missing_df[missing_df['Missing'] > 0])

print(f"\n--- عدد الصفوف المكررة: {df.duplicated().sum()} ---")
```

```python

# ============================================================================
# 3) تنظيف البيانات
# ============================================================================
print("\n🧹 [3] تنظيف البيانات...")

# تحويل التاريخ
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

# تعبئة القيم المفقودة
df['Region'] = df['Region'].fillna('Unknown')
df['Shipping Status'] = df['Shipping Status'].fillna('Unknown')

# تعبئة Age بالوسيط
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)
print(f"   ✓ تم تعبئة Age بالوسيط: {age_median}")

# إزالة المكررات
before = len(df)
df = df.drop_duplicates()
print(f"   ✓ تم حذف {before - len(df)} صف مكرر")

# إزالة الصفوف بكميات أو أسعار غير صحيحة
df = df[(df['Quantity'] > 0) & (df['Unit Price'] > 0)].reset_index(drop=True)
print(f"   ✅ شكل البيانات النهائي: {df.shape}")
```

```python

# ============================================================================
# 4) هندسة الميزات (Feature Engineering)
# ============================================================================
print("\n⚙️ [4] هندسة الميزات...")

df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month_Name'] = df['Order Date'].dt.month_name()
df['Day_of_Week'] = df['Order Date'].dt.day_name()

# فئة عمرية
def age_group(age):
    if age < 25:
        return 'Youth (18-24)'
    elif age < 35:
        return 'Young Adult (25-34)'
    elif age < 50:
        return 'Adult (35-49)'
    elif age < 65:
        return 'Middle Aged (50-64)'
    else:
        return 'Senior (65+)'

df['Age_Group'] = df['Age'].apply(age_group)

# تصنيف قيمة الطلب
def order_size(total):
    if total < 100:
        return 'Small'
    elif total < 500:
        return 'Medium'
    elif total < 2000:
        return 'Large'
    else:
        return 'Extra Large'

df['Order_Size'] = df['Total Price'].apply(order_size)

print("   ✓ تم إضافة: Year, Month, Month_Name, Day_of_Week, Age_Group, Order_Size")
```

```python

# ============================================================================
# 5) التحليل الأحادي (Univariate)
# ============================================================================
print("\n📊 [5] التحليل الأحادي...")

# 5.1 توزيع الجنس
plt.figure(figsize=(8, 6))
gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%',
        colors=COLORS[:2], startangle=90, explode=[0.05]*2)
plt.title('Gender Distribution\nتوزيع الجنس', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('charts/01_gender_distribution.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 01_gender_distribution.png")

# 5.2 توزيع المناطق
plt.figure(figsize=(10, 6))
region_counts = df['Region'].value_counts()
bars = plt.bar(region_counts.index, region_counts.values, color=COLORS[:len(region_counts)])
plt.title('Region Distribution\nتوزيع المناطق', fontsize=14, fontweight='bold')
plt.xlabel('Region')
plt.ylabel('Number of Orders')
for bar, val in zip(bars, region_counts.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
             f'{val}\n({val/len(df)*100:.1f}%)', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/02_region_distribution.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 02_region_distribution.png")

# 5.3 توزيع الفئات
plt.figure(figsize=(10, 6))
cat_counts = df['Category'].value_counts()
bars = plt.bar(cat_counts.index, cat_counts.values, color=COLORS[:len(cat_counts)])
plt.title('Category Distribution\nتوزيع الفئات', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Number of Orders')
for bar, val in zip(bars, cat_counts.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
             f'{val}\n({val/len(df)*100:.1f}%)', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/03_category_distribution.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 03_category_distribution.png")

# 5.4 توزيع المنتجات
plt.figure(figsize=(12, 6))
prod_counts = df['Product Name'].value_counts()
bars = plt.bar(range(len(prod_counts)), prod_counts.values, color=COLORS[:len(prod_counts)])
plt.title('Product Name Distribution\nتوزيع المنتجات', fontsize=14, fontweight='bold')
plt.xlabel('Product')
plt.ylabel('Number of Orders')
plt.xticks(range(len(prod_counts)), prod_counts.index, rotation=20)
for bar, val in zip(bars, prod_counts.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
             f'{val}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/04_product_distribution.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 04_product_distribution.png")

# 5.5 توزيع حالة الشحن
plt.figure(figsize=(8, 6))
status_counts = df['Shipping Status'].value_counts()
bars = plt.bar(status_counts.index, status_counts.values, color=COLORS[:len(status_counts)])
plt.title('Shipping Status Distribution\nتوزيع حالة الشحن', fontsize=14, fontweight='bold')
plt.xlabel('Status')
plt.ylabel('Number of Orders')
for bar, val in zip(bars, status_counts.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
             f'{val}\n({val/len(df)*100:.1f}%)', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/05_shipping_status.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 05_shipping_status.png")

# 5.6 توزيع الأسعار
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].hist(df['Total Price'], bins=30, color=COLORS[0], edgecolor='white', alpha=0.8)
axes[0].set_title('Distribution of Total Price\nتوزيع إجمالي السعر', fontweight='bold')
axes[0].set_xlabel('Total Price ($)')
axes[0].set_ylabel('Frequency')
axes[0].axvline(df['Total Price'].mean(), color='red', linestyle='--',
                label=f'Mean: ${df["Total Price"].mean():.2f}')
axes[0].axvline(df['Total Price'].median(), color='green', linestyle='--',
                label=f'Median: ${df["Total Price"].median():.2f}')
axes[0].legend()

bp = axes[1].boxplot(df['Total Price'], vert=True, patch_artist=True)
bp['boxes'][0].set_facecolor(COLORS[1])
axes[1].set_title('Box Plot of Total Price', fontweight='bold')
axes[1].set_ylabel('Total Price ($)')

plt.tight_layout()
plt.savefig('charts/06_price_distribution.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 06_price_distribution.png")
```

```python


# ============================================================================
# 6) التحليل الثنائي (Bivariate)
# ============================================================================
print("\n📈 [6] التحليل الثنائي...")

# 6.1 الإيرادات حسب الفئة
cat_revenue = df.groupby('Category').agg(
    revenue=('Total Price', 'sum'),
    orders=('Total Price', 'count'),
    avg_price=('Total Price', 'mean')
).sort_values('revenue', ascending=False)

plt.figure(figsize=(10, 6))
bars = plt.bar(cat_revenue.index, cat_revenue['revenue'], color=COLORS[:len(cat_revenue)])
plt.title('Total Revenue by Category\nالإيرادات حسب الفئة', fontsize=14, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Revenue ($)')
for bar, val in zip(bars, cat_revenue['revenue']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1000,
             f'${val:,.0f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/07_revenue_by_category.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 07_revenue_by_category.png")

# 6.2 الإيرادات حسب المنتج
prod_revenue = df.groupby('Product Name').agg(
    revenue=('Total Price', 'sum'),
    quantity=('Quantity', 'sum')
).sort_values('revenue', ascending=False)

plt.figure(figsize=(12, 6))
bars = plt.bar(range(len(prod_revenue)), prod_revenue['revenue'],
               color=COLORS[:len(prod_revenue)])
plt.title('Total Revenue by Product\nالإيرادات حسب المنتج', fontsize=14, fontweight='bold')
plt.xlabel('Product')
plt.ylabel('Revenue ($)')
plt.xticks(range(len(prod_revenue)), prod_revenue.index, rotation=20)
for bar, val in zip(bars, prod_revenue['revenue']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1000,
             f'${val:,.0f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/08_revenue_by_product.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 08_revenue_by_product.png")

# 6.3 الإيرادات حسب المنطقة
region_revenue = df.groupby('Region').agg(
    revenue=('Total Price', 'sum'),
    orders=('Total Price', 'count')
).sort_values('revenue', ascending=False)

plt.figure(figsize=(10, 6))
bars = plt.bar(region_revenue.index, region_revenue['revenue'],
               color=COLORS[:len(region_revenue)])
plt.title('Revenue by Region\nالإيرادات حسب المنطقة', fontsize=14, fontweight='bold')
plt.xlabel('Region')
plt.ylabel('Revenue ($)')
for bar, val, orders in zip(bars, region_revenue['revenue'], region_revenue['orders']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1000,
             f'${val:,.0f}\n({orders} orders)', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('charts/09_revenue_by_region.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 09_revenue_by_region.png")

# 6.4 الجنس × الفئة
gender_cat = pd.crosstab(df['Gender'], df['Category'])
plt.figure(figsize=(10, 6))
gender_cat.plot(kind='bar', ax=plt.gca(), color=COLORS[:len(gender_cat.columns)])
plt.title('Category Preference by Gender\nتفضيل الفئة حسب الجنس', fontsize=14, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Number of Orders')
plt.legend(title='Category')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('charts/10_gender_vs_category.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 10_gender_vs_category.png")

# 6.5 المجموعة العمرية × المنتج
age_product = pd.crosstab(df['Age_Group'], df['Product Name'])
plt.figure(figsize=(14, 7))
age_product.plot(kind='bar', stacked=True, ax=plt.gca(),
                 color=COLORS[:len(age_product.columns)])
plt.title('Product Preference by Age Group\nتفضيل المنتج حسب المجموعة العمرية',
          fontsize=14, fontweight='bold')
plt.xlabel('Age Group')
plt.ylabel('Number of Orders')
plt.legend(title='Product', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig('charts/11_age_vs_product.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 11_age_vs_product.png")
```

```python

# ============================================================================
# 7) التحليل الزمني
# ============================================================================
print("\n⏰ [7] التحليل الزمني...")

# 7.1 المبيعات الشهرية
monthly = df.groupby(['Year', 'Month']).agg(
    revenue=('Total Price', 'sum'),
    orders=('Total Price', 'count')
).reset_index()
monthly['date'] = pd.to_datetime(monthly[['Year', 'Month']].assign(day=1))

fig, axes = plt.subplots(2, 1, figsize=(14, 10))
axes[0].plot(monthly['date'], monthly['revenue'], marker='o',
             linewidth=2, markersize=8, color=COLORS[0])
axes[0].fill_between(monthly['date'], monthly['revenue'], alpha=0.3, color=COLORS[0])
axes[0].set_title('Monthly Revenue Trend\nاتجاه الإيرادات الشهرية', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Revenue ($)')

axes[1].bar(monthly['date'], monthly['orders'], width=20,
            color=COLORS[1], alpha=0.8)
axes[1].set_title('Monthly Orders Count\nعدد الطلبات شهرياً', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Number of Orders')

plt.tight_layout()
plt.savefig('charts/12_monthly_trends.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 12_monthly_trends.png")

# 7.2 المبيعات حسب يوم الأسبوع
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily = df.groupby('Day_of_Week').agg(
    revenue=('Total Price', 'sum'),
    orders=('Total Price', 'count')
).reindex(day_order)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].bar(daily.index, daily['revenue'], color=COLORS[:7])
axes[0].set_title('Revenue by Day of Week', fontweight='bold')
axes[0].set_ylabel('Revenue ($)')
axes[0].tick_params(axis='x', rotation=30)

axes[1].bar(daily.index, daily['orders'], color=COLORS[3:10])
axes[1].set_title('Orders by Day of Week', fontweight='bold')
axes[1].set_ylabel('Number of Orders')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.savefig('charts/13_day_of_week.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 13_day_of_week.png")
```

```python


# ============================================================================
# 8) تحليل متعدد المتغيرات
# ============================================================================
print("\n🔬 [8] تحليل متعدد المتغيرات...")

# 8.1 خريطة الارتباط
numeric_cols = ['Age', 'Unit Price', 'Quantity', 'Total Price', 'Shipping Fee']
corr = df[numeric_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='RdYlBu_r', center=0,
            fmt='.2f', square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8})
plt.title('Correlation Heatmap\nخريطة الارتباط', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('charts/14_correlation.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 14_correlation.png")

# 8.2 المنطقة × حالة الشحن
region_status = pd.crosstab(df['Region'], df['Shipping Status'], normalize='index') * 100
plt.figure(figsize=(12, 6))
region_status.plot(kind='bar', stacked=True, ax=plt.gca(),
                   color=COLORS[:len(region_status.columns)])
plt.title('Shipping Status by Region (%)\nحالة الشحن حسب المنطقة', fontsize=14, fontweight='bold')
plt.xlabel('Region')
plt.ylabel('Percentage (%)')
plt.legend(title='Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('charts/15_region_vs_status.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 15_region_vs_status.png")

# 8.3 العلاقة بين السعر وكمية
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Unit Price', y='Quantity', hue='Category',
                size='Total Price', sizes=(50, 300), palette=COLORS[:3])
plt.title('Unit Price vs Quantity (by Category)\nالسعر مقابل الكمية', fontsize=14, fontweight='bold')
plt.xlabel('Unit Price ($)')
plt.ylabel('Quantity')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('charts/16_price_vs_quantity.png', dpi=100, bbox_inches='tight')
plt.close()
print("   ✓ 16_price_vs_quantity.png")
```

```python


# ============================================================================
# 9) الرؤى الرئيسية
# ============================================================================
print("\n💡 [9] الرؤى الرئيسية...")

insights = [
    f"إجمالي الإيرادات: ${df['Total Price'].sum():,.2f}",
    f"عدد الطلبات: {len(df):,}",
    f"متوسط قيمة الطلب: ${df['Total Price'].mean():.2f}",
    f"أفضل فئة: {cat_revenue.index[0]} (${cat_revenue['revenue'].iloc[0]:,.2f})",
    f"أفضل منتج: {prod_revenue.index[0]} (${prod_revenue['revenue'].iloc[0]:,.2f})",
    f"أفضل منطقة: {region_revenue.index[0]} (${region_revenue['revenue'].iloc[0]:,.2f})",
    f"متوسط عمر العميل: {df['Age'].mean():.1f} سنة",
    f"الجنس الأكثر شراءً: {gender_counts.index[0]} ({gender_counts.iloc[0]/len(df)*100:.1f}%)",
    f"نسبة المرتجعات: {(df['Shipping Status']=='Returned').sum()/len(df)*100:.1f}%",
    f"نسبة الطلبات المسلّمة: {(df['Shipping Status']=='Delivered').sum()/len(df)*100:.1f}%",
    f"متوسط رسوم الشحن: ${df['Shipping Fee'].mean():.2f}",
    f"أكثر شهر مبيعاً: {monthly.loc[monthly['revenue'].idxmax(), 'date'].strftime('%B %Y')}",
    f"أكثر يوم نشاطاً: {daily['orders'].idxmax()} ({int(daily['orders'].max())} طلب)",
    f"متوسط الكمية لكل طلب: {df['Quantity'].mean():.2f}",
]

print("\n" + "=" * 70)
print("📊 الرؤى الرئيسية (Key Insights)")
print("=" * 70)
for i, insight in enumerate(insights, 1):
    print(f"  {i:2d}. {insight}")

# حفظ الرؤى
with open('key_insights.txt', 'w', encoding='utf-8') as f:
    f.write("=" * 70 + "\n")
    f.write("Key Insights - E-commerce Sales EDA\n")
    f.write("=" * 70 + "\n\n")
    for i, insight in enumerate(insights, 1):
        f.write(f"{i:2d}. {insight}\n")
    f.write("\n" + "=" * 70 + "\n")
    f.write(f"Total Records Analyzed: {len(df):,}\n")

# حفظ البيانات النظيفة
df.to_csv('ecommerce_clean.csv', index=False)

print("\n" + "=" * 70)
print(f"✅ تم إنتاج 16 رسم بياني في charts/")
print(f"✅ تم حفظ البيانات النظيفة في ecommerce_clean.csv")
print(f"✅ تم حفظ الرؤى في key_insights.txt")
print("=" * 70)
```

```python
"""
إنشاء تقرير PDF احترافي لتحليل التجارة الإلكترونية
يجمع كل الصور والرؤى في ملف PDF واحد
"""

import os
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch, cm
from reportlab.lib.colors import HexColor, white, black
from reportlab.lib.enums import TA_CENTER, TA_LEFT, TA_JUSTIFY
from reportlab.platypus import (SimpleDocTemplate, Paragraph, Spacer, Image,
                                 PageBreak, Table, TableStyle)

# ============================================================================
# الإعدادات
# ============================================================================
OUTPUT_PDF = 'Ecommerce_Sales_Report.pdf'
CHARTS_DIR = 'charts'
INSIGHTS_FILE = 'key_insights.txt'

# ألوان احترافية
PRIMARY_COLOR = HexColor('#2E86AB')
SECONDARY_COLOR = HexColor('#A23B72')
ACCENT_COLOR = HexColor('#F18F01')
DARK_COLOR = HexColor('#1A1A2E')
LIGHT_COLOR = HexColor('#F8F9FA')

# ============================================================================
# إنشاء الأنماط (Styles)
# ============================================================================
styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    'CustomTitle', parent=styles['Title'], fontSize=28,
    textColor=PRIMARY_COLOR, spaceAfter=30,
    alignment=TA_CENTER, fontName='Helvetica-Bold'
)

subtitle_style = ParagraphStyle(
    'CustomSubtitle', parent=styles['Normal'], fontSize=16,
    textColor=SECONDARY_COLOR, spaceAfter=20,
    alignment=TA_CENTER, fontName='Helvetica'
)

heading_style = ParagraphStyle(
    'CustomHeading', parent=styles['Heading1'], fontSize=20,
    textColor=PRIMARY_COLOR, spaceAfter=15, spaceBefore=20,
    fontName='Helvetica-Bold'
)

subheading_style = ParagraphStyle(
    'CustomSubheading', parent=styles['Heading2'], fontSize=14,
    textColor=SECONDARY_COLOR, spaceAfter=10, spaceBefore=15,
    fontName='Helvetica-Bold'
)

body_style = ParagraphStyle(
    'CustomBody', parent=styles['Normal'], fontSize=11,
    textColor=DARK_COLOR, spaceAfter=10,
    alignment=TA_JUSTIFY, fontName='Helvetica', leading=16
)

caption_style = ParagraphStyle(
    'CustomCaption', parent=styles['Normal'], fontSize=10,
    textColor=ACCENT_COLOR, spaceAfter=20,
    alignment=TA_CENTER, fontName='Helvetica-Oblique'
)

# ============================================================================
# قائمة الرسومات مع عناوينها ووصفها
# ============================================================================
charts_info = [
    ('01_gender_distribution.png', 'Gender Distribution', 'توزيع الجنس',
     'يوضح هذا الرسم نسبة المبيعات بين الذكور والإناث. الذكور يشكّلون النسبة الأعلى.'),
    ('02_region_distribution.png', 'Region Distribution', 'توزيع المناطق',
     'يوضح توزيع الطلبات على المناطق الجغرافية الأربعة.'),
    ('03_category_distribution.png', 'Category Distribution', 'توزيع الفئات',
     'يعرض عدد الطلبات لكل فئة. Electronics هي الأكثر طلباً.'),
    ('04_product_distribution.png', 'Product Distribution', 'توزيع المنتجات',
     'يوضح عدد الطلبات لكل منتج.'),
    ('05_shipping_status.png', 'Shipping Status', 'حالة الشحن',
     'يعرض توزيع حالات الشحن. نسبة المرتجعات تستوجب الاهتمام.'),
    ('06_price_distribution.png', 'Price Distribution', 'توزيع الأسعار',
     'يوضح توزيع إجمالي الأسعار عبر جميع الطلبات.'),
    ('07_revenue_by_category.png', 'Revenue by Category', 'الإيرادات حسب الفئة',
     'يعرض إجمالي الإيرادات لكل فئة. Electronics تحقق الأعلى.'),
    ('08_revenue_by_product.png', 'Revenue by Product', 'الإيرادات حسب المنتج',
     'يوضح إجمالي الإيرادات لكل منتج. Laptop هو الأعلى إيراداً.'),
    ('09_revenue_by_region.png', 'Revenue by Region', 'الإيرادات حسب المنطقة',
     'يعرض إجمالي الإيرادات لكل منطقة. East تحقق الأعلى.'),
    ('10_gender_vs_category.png', 'Gender vs Category', 'الجنس مقابل الفئة',
     'يوضح تفضيلات الفئات بين الجنسين.'),
    ('11_age_vs_product.png', 'Age Group vs Product', 'العمر مقابل المنتج',
     'يعرض تفضيل المنتجات حسب المجموعة العمرية.'),
    ('12_monthly_trends.png', 'Monthly Trends', 'الاتجاهات الشهرية',
     'يوضح تطور الإيرادات وعدد الطلبات شهرياً.'),
    ('13_day_of_week.png', 'Day of Week Analysis', 'تحليل أيام الأسبوع',
     'يعرض المبيعات حسب أيام الأسبوع.'),
    ('14_correlation.png', 'Correlation Heatmap', 'خريطة الارتباط',
     'توضح العلاقات بين المتغيرات الرقمية.'),
    ('15_region_vs_status.png', 'Region vs Shipping Status', 'المنطقة مقابل حالة الشحن',
     'يعرض نسب حالات الشحن في كل منطقة.'),
    ('16_price_vs_quantity.png', 'Price vs Quantity', 'السعر مقابل الكمية',
     'يوضح العلاقة بين سعر الوحدة والكمية المطلوبة.'),
]

# ============================================================================
# بناء التقرير
# ============================================================================
doc = SimpleDocTemplate(
    OUTPUT_PDF, pagesize=A4,
    rightMargin=2*cm, leftMargin=2*cm,
    topMargin=2*cm, bottomMargin=2*cm
)

story = []

# 1) صفحة الغلاف
story.append(Spacer(1, 4*cm))
story.append(Paragraph('E-commerce Sales', title_style))
story.append(Paragraph('Exploratory Data Analysis Report', subtitle_style))
story.append(Spacer(1, 1*cm))

separator = Table([['']], colWidths=[16*cm], rowHeights=[2])
separator.setStyle(TableStyle([('BACKGROUND', (0,0), (-1,-1), PRIMARY_COLOR)]))
story.append(separator)

story.append(Spacer(1, 1*cm))
story.append(Paragraph('تحليل بيانات مبيعات التجارة الإلكترونية', subtitle_style))

story.append(Spacer(1, 3*cm))

info_data = [
    ['Report Type:', 'Exploratory Data Analysis (EDA)'],
    ['Dataset:', 'E-commerce Sales Data'],
    ['Total Records:', '982 transactions'],
    ['Analysis Date:', '2024'],
    ['Tools Used:', 'Python, Pandas, Matplotlib, Seaborn'],
]

info_table = Table(info_data, colWidths=[5*cm, 10*cm])
info_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (0,-1), PRIMARY_COLOR),
    ('TEXTCOLOR', (0,0), (0,-1), white),
    ('TEXTCOLOR', (1,0), (1,-1), DARK_COLOR),
    ('FONTNAME', (0,0), (0,-1), 'Helvetica-Bold'),
    ('FONTNAME', (1,0), (1,-1), 'Helvetica'),
    ('FONTSIZE', (0,0), (-1,-1), 11),
    ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
    ('TOPPADDING', (0,0), (-1,-1), 8),
    ('BOTTOMPADDING', (0,0), (-1,-1), 8),
    ('LEFTPADDING', (0,0), (-1,-1), 10),
    ('GRID', (0,0), (-1,-1), 0.5, white),
]))
story.append(info_table)
story.append(PageBreak())

# 2) Executive Summary
story.append(Paragraph('1. Executive Summary', heading_style))
story.append(Paragraph(
    'This report presents a comprehensive Exploratory Data Analysis of e-commerce sales '
    'data containing 982 transactions. The analysis covers customer demographics, product '
    'preferences, geographic distribution, temporal patterns, and shipping performance. '
    'Key findings: Electronics dominates sales revenue, Laptop is the top product, East '
    'region generates highest revenue, and November 2023 was the peak sales month. '
    'The return rate of 30.4% requires immediate attention.', body_style))
story.append(PageBreak())

# 3) Introduction
story.append(Paragraph('2. Introduction', heading_style))
story.append(Paragraph(
    'Exploratory Data Analysis (EDA) is a critical first step in any data science project. '
    'This project analyzes real e-commerce sales data to answer key business questions: '
    'What are the most popular products? Which regions generate the most revenue? '
    'How do customer demographics affect purchasing behavior? What are the temporal patterns?',
    body_style))
story.append(PageBreak())

# 4) جميع الرسومات مع شرحها
story.append(Paragraph('3. Visual Analysis', heading_style))
story.append(Paragraph(
    'This section presents all visualizations organized by analysis type.', body_style))
story.append(Spacer(1, 0.5*cm))

for img_file, en_title, ar_title, description in charts_info:
    img_path = os.path.join(CHARTS_DIR, img_file)
    if os.path.exists(img_path):
        story.append(Paragraph(f'{en_title}', subheading_style))
        story.append(Paragraph(f'<i>{ar_title}</i>', caption_style))
        img = Image(img_path, width=16*cm, height=8*cm)
        story.append(img)
        story.append(Spacer(1, 0.3*cm))
        story.append(Paragraph(description, body_style))
        story.append(Spacer(1, 0.5*cm))

story.append(PageBreak())

# 5) Key Insights
story.append(Paragraph('4. Key Insights', heading_style))
story.append(Paragraph('The following key insights were derived from the analysis:', body_style))
story.append(Spacer(1, 0.5*cm))

# قراءة الرؤى من الملف
insights_list = []
if os.path.exists(INSIGHTS_FILE):
    with open(INSIGHTS_FILE, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line and line[0].isdigit():
                insights_list.append(line)

if insights_list:
    insights_data = [['#', 'Insight']]
    for line in insights_list:
        parts = line.split('. ', 1)
        if len(parts) == 2:
            insights_data.append([parts[0], parts[1]])

    insights_table = Table(insights_data, colWidths=[1.5*cm, 14.5*cm])
    insights_table.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), PRIMARY_COLOR),
        ('TEXTCOLOR', (0,0), (-1,0), white),
        ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
        ('FONTSIZE', (0,0), (-1,-1), 10),
        ('ALIGN', (0,0), (0,-1), 'CENTER'),
        ('VALIGN', (0,0), (-1,-1), 'MIDDLE'),
        ('TOPPADDING', (0,0), (-1,-1), 8),
        ('BOTTOMPADDING', (0,0), (-1,-1), 8),
        ('GRID', (0,0), (-1,-1), 0.5, HexColor('#CCCCCC')),
        ('ROWBACKGROUNDS', (0,1), (-1,-1), [white, LIGHT_COLOR]),
    ]))
    story.append(insights_table)

story.append(PageBreak())

# 6) Recommendations
story.append(Paragraph('5. Recommendations', heading_style))

recommendations = [
    ('1. Reduce Return Rate',
     'The return rate of 30.4% is significantly high. Investigate product quality issues '
     'and improve customer service.'),
    ('2. Focus on Electronics',
     'Electronics generates the highest revenue ($1,179,600). Expand this category.'),
    ('3. Target East Region',
     'East region generates the highest revenue. Develop targeted marketing campaigns.'),
    ('4. November Sales Strategy',
     'November was the peak sales month. Prepare inventory for November each year.'),
    ('5. Tuesday Promotions',
     'Tuesday has the highest orders. Schedule promotions on Tuesdays.'),
    ('6. Age-Based Marketing',
     'Different age groups prefer different products. Develop targeted campaigns.'),
]

for title, desc in recommendations:
    story.append(Paragraph(f'<b>{title}</b>', subheading_style))
    story.append(Paragraph(desc, body_style))
    story.append(Spacer(1, 0.3*cm))

# خاتمة
story.append(Spacer(1, 2*cm))
separator2 = Table([['']], colWidths=[16*cm], rowHeights=[2])
separator2.setStyle(TableStyle([('BACKGROUND', (0,0), (-1,-1), PRIMARY_COLOR)]))
story.append(separator2)
story.append(Spacer(1, 0.5*cm))
story.append(Paragraph('End of Report', ParagraphStyle(
    'EndStyle', parent=body_style, fontSize=14,
    textColor=PRIMARY_COLOR, alignment=TA_CENTER,
    fontName='Helvetica-Bold')))

# بناء الـ PDF
doc.build(story)
print(f"✅ تم إنشاء التقرير: {OUTPUT_PDF}")
```
